```python
# === 1. IMPORTS ===
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# %matplotlib inline # This is a Jupyter magic command, keep if in Jupyter
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import (accuracy_score, balanced_accuracy_score,
                             classification_report, confusion_matrix,
                             precision_score, recall_score, f1_score)
import pennylane as qml
from pennylane.templates.embeddings import AngleEmbedding
from pennylane.templates.layers import StronglyEntanglingLayers
from glob import glob
import time
import seaborn as sns
from imblearn.over_sampling import SMOTE
from collections import Counter
from tqdm import tqdm
import traceback
import tensorflow as tf

# === 2. CONFIGURATION ===
# Dataset configuration
image_dir = "/content/drive/MyDrive/archive/data"
classes = ["cloudy", "foggy", "rainy", "snowy", "sunny"]
image_size = (64, 64)
USE_COLOR = True

# Model parameters
random_state = 42
n_qubits = 12
pca_variance = 0.95
max_quantum_samples = 200
qsvm_layers = 2

# Experiment parameters
sample_sizes = [50, 100, 200, 300]  # Sample sizes to test
test_size = 0.2  # Test set proportion

# Initialize quantum device
dev = qml.device("default.qubit", wires=n_qubits)

# === 3. DATA LOADING ===
def load_images(base_path, class_names, target_size=(64, 64), max_per_class=500): # Adjusted default target_size to match config
    """Load and preprocess images from directory"""
    X, y = [], []
    preview_images = {cls: [] for cls in class_names}

    for label_idx, class_name in enumerate(class_names):
        class_path = os.path.join(base_path, class_name)
        if not os.path.exists(class_path):
            print(f"Warning: Missing directory {class_path}")
            continue

        image_paths = glob(os.path.join(class_path, "*.jpg"))[:max_per_class]
        print(f"Loading {len(image_paths)} images for {class_name}")

        for img_path in tqdm(image_paths, desc=class_name):
            try:
                img = cv2.imread(img_path)
                if img is None:
                    print(f"Warning: Could not read image {img_path}, skipping.")
                    continue

                if USE_COLOR:
                    processed_img = cv2.resize(img, target_size)
                    if processed_img.shape[-1] != 3: # Ensure 3 channels if color
                        print(f"Warning: Image {img_path} not in 3 channels despite USE_COLOR=True. Shape: {processed_img.shape}")
                        # Potentially convert to BGR if it's single channel interpreted as color
                        if len(processed_img.shape) == 2: # Grayscale
                           processed_img = cv2.cvtColor(processed_img, cv2.COLOR_GRAY2BGR)
                        # Add other specific conversions if necessary, or skip
                        # For now, we assume cv2.imread and resize provides 3 channels if original is color
                else:
                    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
                    processed_img = cv2.resize(gray, target_size)

                X.append((processed_img.astype(np.float32) / 255.0).flatten())
                y.append(label_idx)

                if len(preview_images[class_name]) < 5:
                    preview_images[class_name].append(cv2.resize(img, (250,250))) # Original img for preview

            except Exception as e:
                print(f"Error processing {img_path}: {str(e)}")

    return np.array(X), np.array(y), preview_images

# === 4. DATA VISUALIZATION ===
def show_class_distribution(y_data, class_names_list):
    """Plot distribution of classes"""
    counts = np.bincount(y_data, minlength=len(class_names_list))
    plt.figure(figsize=(10,5))
    plt.bar(class_names_list, counts)
    plt.title("Class Distribution")
    plt.xlabel("Class")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()

def show_sample_images(preview_dict, class_names_list):
    """Display sample images from each class"""
    plt.figure(figsize=(15, 10)) # Adjusted size for better layout
    num_preview = 5
    for i, cls in enumerate(class_names_list):
        if cls not in preview_dict or not preview_dict[cls]:
            print(f"No preview images for class: {cls}")
            continue
        for j, img in enumerate(preview_dict[cls][:num_preview]):
            plt.subplot(len(class_names_list), num_preview, i * num_preview + j + 1)
            plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            if j == 0:
                plt.title(f"{cls}")
            plt.axis('off')
    plt.tight_layout()
    plt.show()

# === 5. PREPROCESSING ===
# Define the CNN feature extractor model
def create_cnn_feature_extractor():
    base_model = tf.keras.applications.ResNet50(
        weights="imagenet",
        include_top=False,
        input_shape=(image_size[0], image_size[1], 3) # ResNet50 expects 3 channels
    )
    base_model.trainable = False

    inputs = base_model.input
    x = base_model.output
    x = tf.keras.layers.GlobalAveragePooling2D(name="global_average_pooling")(x)
    outputs = x

    features_extractor_model = tf.keras.Model(inputs=inputs, outputs=outputs, name="cnn_feature_extractor")
    print("CNN Feature Extractor Model Summary:")
    features_extractor_model.summary()
    return features_extractor_model

# Create the feature extractor instance globally
feature_extractor = create_cnn_feature_extractor()

def extract_features_cnn(X_pixel_data):
    """Extract features using the pre-trained CNN.
    X_pixel_data is expected to be flattened image data.
    """
    print(f"Inside extract_features_cnn - Input X_pixel_data shape: {X_pixel_data.shape}")

    # Determine current channels based on USE_COLOR config
    # This config should align with how load_images prepared the data.
    data_channels = 3 if USE_COLOR else 1

    # Reshape flattened images back to image format (batch, height, width, channels)
    try:
        images_for_cnn = X_pixel_data.reshape((-1, image_size[0], image_size[1], data_channels))
    except ValueError as e:
        print(f"Error reshaping X_pixel_data: {e}")
        print(f"Expected flattened features per image: {image_size[0]*image_size[1]*data_channels}")
        print(f"Actual features per image in X_pixel_data: {X_pixel_data.shape[1]}")
        raise

    print(f"Reshaped images_for_cnn shape: {images_for_cnn.shape}")

    model_expected_channels = feature_extractor.input_shape[-1] # Should be 3 for ResNet50

    if images_for_cnn.shape[-1] != model_expected_channels:
        if images_for_cnn.shape[-1] == 1 and model_expected_channels == 3:
            print(f"Converting {images_for_cnn.shape[-1]}-channel images to {model_expected_channels}-channel for model.")
            images_for_cnn = np.repeat(images_for_cnn, model_expected_channels, axis=-1)
            print(f"Shape after channel repeat: {images_for_cnn.shape}")
        else:
            raise ValueError(
                f"Channel mismatch: Input images have {images_for_cnn.shape[-1]} channels, "
                f"but the CNN model's input layer expects {model_expected_channels} channels."
            )

    # Ensure the data is C-contiguous before passing to TensorFlow
    if not images_for_cnn.flags['C_CONTIGUOUS']:
        print("Warning: images_for_cnn is not C-contiguous. Making a C-contiguous copy.")
        images_for_cnn = np.ascontiguousarray(images_for_cnn)

    print(f"Final shape of images_for_cnn before predict: {images_for_cnn.shape}")

    # Explicitly set batch_size for predict
    X_cnn_features = feature_extractor.predict(images_for_cnn, batch_size=32) # Default TF batch_size is 32

    print(f"Shape of X_cnn_features AFTER predict: {X_cnn_features.shape}") # CRITICAL DEBUG PRINT

    return X_cnn_features

def preprocess_data(X_input, y_input):
    """Apply preprocessing pipeline"""
    print("\nPreprocessing data...")
    print(f"Initial X shape: {X_input.shape}, Initial y shape: {y_input.shape}")

    # 1. Handle class imbalance with SMOTE
    print("Balancing classes with SMOTE...")
    smote = SMOTE(random_state=random_state)
    X_bal, y_bal = smote.fit_resample(X_input, y_input)
    print(f"Class distribution after SMOTE: {Counter(y_bal)}")
    print(f"X_bal shape after SMOTE: {X_bal.shape}, y_bal shape after SMOTE: {y_bal.shape}")

    # 2. Feature Extraction using CNN
    print("Extracting features using CNN...")
    X_cnn_features = extract_features_cnn(X_bal) # Pass balanced pixel data
    print(f"X_cnn_features shape after extract_features_cnn: {X_cnn_features.shape}")

    # 3. Standard scaling (applied to the extracted CNN features)
    print("Applying StandardScaler...")
    scaler = StandardScaler()
    X_scaled_cnn_features = scaler.fit_transform(X_cnn_features)
    print(f"X_scaled_cnn_features shape after scaling: {X_scaled_cnn_features.shape}")

    # 4. Dimensionality reduction with PCA (applied to scaled CNN features)
    print(f"Applying PCA (retaining {pca_variance*100}% variance)...")
    n_samples_for_pca, n_features_for_pca = X_scaled_cnn_features.shape

    current_pca_variance = pca_variance
    if isinstance(pca_variance, float) and (pca_variance > 1.0 or pca_variance <=0):
        print(f"Warning: pca_variance ({pca_variance}) is out of (0,1] range. Setting to 0.95")
        current_pca_variance = 0.95
    # n_components for PCA should be min(n_samples, n_features) if it's an int.
    # If float, it's variance to retain. PCA lib handles if requested components > available.

    pca = PCA(n_components=current_pca_variance, random_state=random_state)
    X_pca_cnn_features = pca.fit_transform(X_scaled_cnn_features)
    print(f"Reduced from {X_scaled_cnn_features.shape[1]} (CNN features) to {X_pca_cnn_features.shape[1]} PCA components.")
    print(f"X_pca_cnn_features shape after PCA: {X_pca_cnn_features.shape}")

    if n_qubits > X_pca_cnn_features.shape[1]:
        print(f"Warning: n_qubits ({n_qubits}) is greater than the number of PCA components ({X_pca_cnn_features.shape[1]}). "
              f"This may lead to errors in AngleEmbedding unless features are sliced. "
              f"Consider setting n_qubits <= {X_pca_cnn_features.shape[1]}.")

    return X_scaled_cnn_features, X_pca_cnn_features, y_bal, scaler, pca

# === 6. CLASSICAL SVM IMPLEMENTATION ===
class ClassicalSVM:
    def __init__(self, kernel='rbf', C=1.0, gamma='scale'):
        self.kernel = kernel
        self.C = C
        self.gamma = gamma
        self.model = SVC(kernel=kernel, C=C, gamma=gamma,
                         probability=True, random_state=random_state)

    def fit(self, X, y):
        self.model.fit(X, y)
        return self

    def predict(self, X):
        return self.model.predict(X)

    def predict_proba(self, X):
        return self.model.predict_proba(X)

    def decision_function(self, X):
        return self.model.decision_function(X)

# === 7. QUANTUM SVM IMPLEMENTATION ===
class QuantumSVM:
    def __init__(self, device, n_qubits_val, max_samples_val=100, n_layers_val=3, C_val=1.0): # Renamed args to avoid conflict
        self.device = device
        self.n_qubits = n_qubits_val
        self.max_samples = max_samples_val
        self.n_layers = n_layers_val
        self.C = C_val

        self.weights = np.random.rand(self.n_layers, self.n_qubits, 3) * (2 * np.pi) # Initialize weights better


        fixed_n_qubits = self.n_qubits

        @qml.qnode(self.device, interface="autograd")
        def quantum_circuit_kernel(x1, x2, weights_param):

            # Pad with zeros if features are fewer than qubits
            len_x1 = len(x1)
            len_x2 = len(x2)

            features_x1 = np.zeros(fixed_n_qubits)
            features_x2 = np.zeros(fixed_n_qubits)

            features_x1[:min(len_x1, fixed_n_qubits)] = x1[:min(len_x1, fixed_n_qubits)]
            features_x2[:min(len_x2, fixed_n_qubits)] = x2[:min(len_x2, fixed_n_qubits)]

            AngleEmbedding(features_x1, wires=range(fixed_n_qubits))
            qml.adjoint(AngleEmbedding(features_x2, wires=range(fixed_n_qubits))) # Adjoint for fidelity-like kernel
            StronglyEntanglingLayers(weights_param, wires=range(fixed_n_qubits))
            return qml.expval(qml.PauliZ(0)) # Example: expectation value, can be other measurements for kernel

        self.quantum_circuit_kernel_func = quantum_circuit_kernel # Store the qnode
        self.svm = None
        self.X_train_subset_for_kernel = None # To store the X used for kernel matrix computation

    def get_params(self, deep=True):
        return {"device": self.device, "n_qubits_val": self.n_qubits,
                "max_samples_val": self.max_samples, "n_layers_val": self.n_layers, "C_val": self.C}

    def set_params(self, **parameters):
        for parameter, value in parameters.items():
            setattr(self, parameter, value)

        return self

    def quantum_kernel_matrix_entry(self, x1_vec, x2_vec):
        """Compute a single quantum kernel value K(x1, x2)"""

        try:
            # Kernel value needs to be scalar and ideally normalized or well-behaved
            val = self.quantum_circuit_kernel_func(x1_vec, x2_vec, self.weights)
            return (val + 1) / 2 # Example normalization if expval is [-1, 1]
        except Exception as e:
            print(f"Error in quantum_kernel_matrix_entry: {e} with x1={x1_vec[:5]}, x2={x2_vec[:5]}") # Print first 5 features
            return 0.0

    def fit(self, X, y):
        """Train the quantum SVM"""
        # Subsample if data is larger than max_samples
        if len(X) > self.max_samples:
            idx = np.random.choice(len(X), self.max_samples, replace=False)
            X_subset, y_subset = X[idx], y[idx]
            print(f"QSVM Training: Subsampled from {len(X)} to {len(X_subset)} samples.")
        else:
            X_subset, y_subset = X, y

        self.X_train_subset_for_kernel = X_subset # Store the subset used for kernel computation

        print("Computing quantum kernel matrix for QSVM fit...")
        num_train_samples = len(X_subset)
        K_train = np.zeros((num_train_samples, num_train_samples))
        for i in tqdm(range(num_train_samples), desc="Kernel rows (training)"):
            for j in range(i, num_train_samples): # Exploit symmetry K[i,j] = K[j,i]
                k_val = self.quantum_kernel_matrix_entry(X_subset[i], X_subset[j])
                K_train[i, j] = k_val
                K_train[j, i] = k_val # Symmetric kernel

        self.svm = SVC(kernel='precomputed', C=self.C, probability=True, random_state=random_state) # enable probability
        self.svm.fit(K_train, y_subset)
        return self

    def _compute_test_kernel_matrix(self, X_test):
        if self.X_train_subset_for_kernel is None:
            raise RuntimeError("QSVM not trained yet or X_train_subset_for_kernel not set.")

        num_test_samples = len(X_test)
        num_train_support_samples = len(self.X_train_subset_for_kernel)

        K_test = np.zeros((num_test_samples, num_train_support_samples))
        print("Computing quantum kernel matrix for QSVM predict/decision_function...")
        for i in tqdm(range(num_test_samples), desc="Kernel rows (testing)"):
            for j in range(num_train_support_samples):
                K_test[i, j] = self.quantum_kernel_matrix_entry(X_test[i], self.X_train_subset_for_kernel[j])
        return K_test

    def decision_function(self, X):
        if self.svm is None:
            raise RuntimeError("QSVM model not trained yet (svm attribute is None).")
        K_test_vs_train = self._compute_test_kernel_matrix(X)
        return self.svm.decision_function(K_test_vs_train)

    def predict_proba(self, X):
        if self.svm is None:
            raise RuntimeError("QSVM model not trained yet (svm attribute is None).")
        if not hasattr(self.svm, 'predict_proba'):
             # Fallback if underlying SVC was not fit with probability=True (should be set in fit)
            print("Warning: predict_proba called on QSVM where underlying SVC might not support it. Using decision_function scores.")
            scores = self.decision_function(X)
            if len(self.svm.classes_) == 2: # Binary case
                probs = 1 / (1 + np.exp(-scores)) # Sigmoid
                return np.vstack([1 - probs, probs]).T
            else:
                raise NotImplementedError("Fallback predict_proba for multiclass QSVM based on scores is not fully implemented here.")

        K_test_vs_train = self._compute_test_kernel_matrix(X)
        return self.svm.predict_proba(K_test_vs_train)

    def predict(self, X):
        if self.svm is None:
            raise RuntimeError("QSVM model not trained yet (svm attribute is None).")
        K_test_vs_train = self._compute_test_kernel_matrix(X)
        return self.svm.predict(K_test_vs_train)

# === 8. MODEL EVALUATION ===
def evaluate_model(model_obj, X_train_data, X_test_data, y_train_data, y_test_data, model_id_name, current_sample_size):
    """Evaluate model and return metrics"""
    print(f"\nEvaluating {model_id_name} (n={current_sample_size})...")

    start_train = time.time()
    model_obj.fit(X_train_data, y_train_data)
    train_time_val = time.time() - start_train

    start_predict = time.time()
    y_pred_vals = model_obj.predict(X_test_data)
    predict_time_val = time.time() - start_predict

    metrics_dict = {
        'accuracy': accuracy_score(y_test_data, y_pred_vals),
        'balanced_accuracy': balanced_accuracy_score(y_test_data, y_pred_vals),
        'precision_macro': precision_score(y_test_data, y_pred_vals, average='macro', zero_division=0),
        'recall_macro': recall_score(y_test_data, y_pred_vals, average='macro', zero_division=0),
        'f1_macro': f1_score(y_test_data, y_pred_vals, average='macro', zero_division=0),
        'train_time': train_time_val,
        'predict_time': predict_time_val
    }

    print(f"Metrics for {model_id_name} (n={current_sample_size}):")
    for k, v in metrics_dict.items():
        print(f"  {k}: {v:.4f}")

    plt.figure(figsize=(8,6))
    cm = confusion_matrix(y_test_data, y_pred_vals, labels=np.arange(len(classes))) # Ensure all classes are considered
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=classes, yticklabels=classes)
    plt.title(f"{model_id_name} Confusion Matrix\n(Sample Size: {current_sample_size})")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.show()

    print("\nClassification Report:")
    print(classification_report(y_test_data, y_pred_vals, target_names=classes, zero_division=0))

    return metrics_dict

# === 9. SINGLE IMAGE TESTING ===
def test_single_image(model_obj, image_path_str, scaler_transform, pca_transform=None, model_type_str='classical'):
    """Test model on a single image with visualization"""
    try:
        img_orig = cv2.imread(image_path_str)
        if img_orig is None:
            print(f"Could not load image: {image_path_str}")
            return

        # Preprocess image similar to training data
        if USE_COLOR:
            processed_display = cv2.resize(img_orig, image_size) # For display if needed
            # For feature extraction, needs to match CNN input if applicable
            img_for_cnn_flat = (processed_display.astype(np.float32)/255.0).flatten().reshape(1,-1)
        else:
            gray_img = cv2.cvtColor(img_orig, cv2.COLOR_BGR2GRAY)
            processed_display = cv2.resize(gray_img, image_size)
            img_for_cnn_flat = (processed_display.astype(np.float32)/255.0).flatten().reshape(1,-1)


        single_img_cnn_features = extract_features_cnn(img_for_cnn_flat) # Shape (1, 2048)

        # Step 2: Scale CNN features
        single_img_scaled_features = scaler_transform.transform(single_img_cnn_features) # Shape (1, 2048)

        if model_type_str == 'quantum':
            if pca_transform is None:
                raise ValueError("PCA transformer must be provided for quantum model type.")
            # Step 3: PCA on scaled CNN features
            img_final_features = pca_transform.transform(single_img_scaled_features) # Shape (1, n_pca_components)
        else: # Classical model uses scaled CNN features
            img_final_features = single_img_scaled_features

        pred_idx = model_obj.predict(img_final_features)[0]
        pred_class_name = classes[pred_idx]

        plt.figure(figsize=(12, 5))
        plt.subplot(1, 2, 1)
        plt.imshow(cv2.cvtColor(cv2.resize(img_orig, (250,250)), cv2.COLOR_BGR2RGB))
        plt.title("Original Image")
        plt.axis('off')

        plt.subplot(1, 2, 2)
        bar_title = f"Prediction: {pred_class_name}"
        try:
            if hasattr(model_obj, 'predict_proba'):
                probas = model_obj.predict_proba(img_final_features)[0]
                # Ensure probas has one entry per class
                if len(probas) != len(classes) and hasattr(model_obj, 'classes_'):
                     # OneVsRestClassifier might return probs only for fitted classes
                     # Reconstruct full probability array if necessary
                     temp_probas = np.zeros(len(classes))
                     fitted_classes_indices = model_obj.classes_ # if model is from sklearn (like SVC, or OneVsRest's estimator)
                     for i, cls_idx in enumerate(fitted_classes_indices):
                         if cls_idx < len(classes): # Make sure class index is valid for `classes` list
                            temp_probas[cls_idx] = probas[i]
                     probas = temp_probas

                if len(probas) == len(classes):
                    plt.bar(classes, probas)
                    plt.title(f"Class Probabilities\n{bar_title}")
                    plt.ylim(0, 1)
                else: # Fallback if probas don't match class count
                    plt.text(0.5, 0.5, bar_title, ha='center', va='center', fontsize=12)
                    print(f"Probabilities array length ({len(probas)}) doesn't match number of classes ({len(classes)}). Displaying prediction only.")

            elif hasattr(model_obj, 'decision_function'):
                scores = model_obj.decision_function(img_final_features)[0]
                # Handling scores for multiclass can be complex to visualize as probabilities
                # For OneVsRest, scores shape is (n_samples, n_classes)
                if len(scores) == len(classes):
                     plt.bar(classes, scores) # Raw scores might not be normalized
                     plt.title(f"Decision Scores\n{bar_title}")
                else:
                    plt.text(0.5, 0.5, bar_title, ha='center', va='center', fontsize=12)
                    print(f"Scores array length ({len(scores)}) doesn't match number of classes ({len(classes)}). Displaying prediction only.")
            else:
                plt.text(0.5, 0.5, bar_title, ha='center', va='center', fontsize=12)

            plt.xticks(rotation=45)
            plt.ylabel("Probability / Score")

        except Exception as viz_err:
             print(f"Error during visualization of probabilities/scores: {viz_err}")
             plt.text(0.5, 0.5, bar_title, ha='center', va='center', fontsize=12) # Fallback

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"Error testing single image {image_path_str}: {e}")
        traceback.print_exc()

# === 10. MAIN EXPERIMENT ===
def run_experiment():
    """Main experiment workflow"""
    print("\n=== LOADING DATA ===")
    # X_pixels_flat will be (N, H*W*C), y_labels will be (N,)
    X_pixels_flat, y_labels, preview_dict_loaded = load_images(image_dir, classes, target_size=image_size, max_per_class=500)
    if X_pixels_flat.size == 0:
        print("No data loaded. Exiting experiment.")
        return

    show_class_distribution(y_labels, classes)
    show_sample_images(preview_dict_loaded, classes)

    print("\n=== PREPROCESSING DATA (once for all experiments) ===")
    # Returns:
    # X_scaled_cnn_feats: CNN features, then scaled. For classical SVMs.
    # X_pca_cnn_feats: CNN features, scaled, then PCA. For Quantum SVMs.
    # y_balanced_labels: Labels after SMOTE, corresponding to the X data.
    # scaler_fitted: The fitted StandardScaler object.
    # pca_fitted: The fitted PCA object.
    X_scaled_cnn_feats, X_pca_cnn_feats, y_balanced_labels, scaler_fitted, pca_fitted = preprocess_data(X_pixels_flat, y_labels)

    print(f"\n--- Shape checks after preprocessing ---")
    print(f"X_scaled_cnn_feats shape: {X_scaled_cnn_feats.shape}")
    print(f"X_pca_cnn_feats shape: {X_pca_cnn_feats.shape}")
    print(f"y_balanced_labels shape: {y_balanced_labels.shape}")
    print(f"-------------------------------------\n")

    # Check for consistency that was causing the error
    if X_scaled_cnn_feats.shape[0] != y_balanced_labels.shape[0]:
        print(f"FATAL ERROR: Mismatch in samples after preprocessing!")
        print(f"X_scaled_cnn_feats samples: {X_scaled_cnn_feats.shape[0]}")
        print(f"y_balanced_labels samples: {y_balanced_labels.shape[0]}")
        return # Stop experiment
    if X_pca_cnn_feats.shape[0] != y_balanced_labels.shape[0]:
        print(f"FATAL ERROR: Mismatch in samples after preprocessing (PCA data)!")
        print(f"X_pca_cnn_feats samples: {X_pca_cnn_feats.shape[0]}")
        print(f"y_balanced_labels samples: {y_balanced_labels.shape[0]}")
        return # Stop experiment


    print("\n=== SPLITTING DATA (into full train/test sets) ===")
    # For classical models (using scaled CNN features)
    X_train_classical, X_test_classical, y_train_classical, y_test_classical = train_test_split(
        X_scaled_cnn_feats, y_balanced_labels,
        test_size=test_size,
        stratify=y_balanced_labels,
        random_state=random_state
    )
    print(f"Classical train shapes: X-{X_train_classical.shape}, y-{y_train_classical.shape}")
    print(f"Classical test shapes: X-{X_test_classical.shape}, y-{y_test_classical.shape}")

    # For quantum models (using PCA of scaled CNN features)
    # Use the same y_balanced_labels for splitting, to ensure y_train_q and y_test_q are same as classical
    X_train_q, X_test_q, y_train_q, y_test_q = train_test_split(
        X_pca_cnn_feats, y_balanced_labels, # Using X_pca_cnn_feats here
        test_size=test_size,
        stratify=y_balanced_labels,
        random_state=random_state # Same random state ensures same split indices for y
    )
    # Verify that y_train_q is same as y_train_classical due to same stratify and random_state
    if not np.array_equal(y_train_classical, y_train_q) or not np.array_equal(y_test_classical, y_test_q):
        print("Warning: y_train/y_test for classical and quantum differ despite same random state. Check split logic.")
        # Fallback to ensure y is consistent if there's an issue, though it should be fine.
        y_train_q = y_train_classical
        y_test_q = y_test_classical


    print(f"Quantum train shapes: X-{X_train_q.shape}, y-{y_train_q.shape}")
    print(f"Quantum test shapes: X-{X_test_q.shape}, y-{y_test_q.shape}")

    models_to_compare = [
        {
            "name": "Linear SVM", "model_builder": lambda: ClassicalSVM(kernel='linear'), # Use lambda to create fresh model
            "param_grid": {'C': [0.1, 1, 10]}, "is_quantum": False
        },
        {
            "name": "RBF SVM", "model_builder": lambda: ClassicalSVM(kernel='rbf'),
            "param_grid": {'C': [0.1, 1, 10], 'gamma': ['scale', 0.1, 0.01]}, "is_quantum": False
        },
        {
            "name": "Quantum SVM",
            "model_builder": lambda: OneVsRestClassifier( # OneVsRest for multi-class QSVM
                QuantumSVM(dev, n_qubits, max_samples_val=max_quantum_samples, n_layers_val=qsvm_layers, C_val=1.0),
                n_jobs=1 # Parallelism for OvR, set to 1 if QSVM part is heavy or has issues with multi-threading
            ),
            "param_grid": None, # GridSearchCV for QSVM parameters can be very slow; usually tune C for SVC or kernel params
            "is_quantum": True
        }
    ]

    results_list = []
    best_models_dict = {} # To store the best performing model for each type for single image test

    for s_size in sample_sizes:
        print(f"\n=== EXPERIMENTING WITH SAMPLE SIZE: {s_size} ===")

        # Create subset from the full training data
        # For classical models
        num_classical_train_samples = len(y_train_classical)
        if s_size < num_classical_train_samples:
            # Stratified subsampling if possible, or random
            # Using train_test_split for stratified subsampling:
            _, X_sub_classical, _, y_sub_classical = train_test_split(
                X_train_classical, y_train_classical,
                test_size=(num_classical_train_samples - s_size)/num_classical_train_samples if num_classical_train_samples > s_size else 0, # fraction to discard
                stratify=y_train_classical,
                random_state=random_state
            )
            print(f"Subsampled classical training data to {len(y_sub_classical)} samples.")
        else:
            X_sub_classical, y_sub_classical = X_train_classical, y_train_classical
            print(f"Using full classical training data: {len(y_sub_classical)} samples (sample_size {s_size} >= available).")

        # For quantum models (from X_train_q, y_train_q)
        num_q_train_samples = len(y_train_q)
        if s_size < num_q_train_samples:
             _, X_sub_q, _, y_sub_q = train_test_split(
                X_train_q, y_train_q,
                test_size=(num_q_train_samples - s_size)/num_q_train_samples if num_q_train_samples > s_size else 0,
                stratify=y_train_q,
                random_state=random_state
            )
             print(f"Subsampled quantum training data to {len(y_sub_q)} samples.")
        else:
            X_sub_q, y_sub_q = X_train_q, y_train_q
            print(f"Using full quantum training data: {len(y_sub_q)} samples (sample_size {s_size} >= available).")

        for config in models_to_compare:
            model_name_str = config["name"]
            is_q_model = config["is_quantum"]

            print(f"\n--- Training {model_name_str} with {s_size} samples ---")

            current_model_instance = config["model_builder"]() # Create a fresh model instance

            if is_q_model:
                X_train_current, y_train_current = X_sub_q, y_sub_q
                X_test_current = X_test_q
                y_test_current = y_test_q # y_test_q is same as y_test_classical
            else:
                X_train_current, y_train_current = X_sub_classical, y_sub_classical
                X_test_current = X_test_classical
                y_test_current = y_test_classical

            # Handle GridSearchCV for classical SVMs
            # For QSVM, param_grid is None (manual params or too slow for grid search here)
            final_model_to_eval = current_model_instance
            if config["param_grid"] and not is_q_model: # Only for classical SVM with grid
                print(f"Performing GridSearchCV for {model_name_str}...")
                # Note: current_model_instance is ClassicalSVM, its .model is the SVC
                grid_search = GridSearchCV(
                    current_model_instance.model, # Pass the underlying SVC model
                    config["param_grid"],
                    cv=3, scoring='accuracy', n_jobs=-1, verbose=1
                )
                grid_search.fit(X_train_current, y_train_current)
                best_params_found = grid_search.best_params_
                print(f"Best params for {model_name_str}: {best_params_found}")

                # Create a new ClassicalSVM instance with best params
                final_model_to_eval = ClassicalSVM( # This is our wrapper
                    kernel=current_model_instance.kernel, # from original config
                    C=best_params_found.get('C', 1.0),
                    gamma=best_params_found.get('gamma', 'scale')
                )
                # No need to fit again here, evaluate_model will do the fit on the final_model_to_eval

            # Evaluate the configured (or grid-searched) model
            metrics_data = evaluate_model(
                final_model_to_eval,
                X_train_current, X_test_current,
                y_train_current, y_test_current, # y_test_classical and y_test_q should be identical
                model_name_str, s_size
            )

            results_list.append({
                'model': model_name_str, 'sample_size': s_size, **metrics_data
            })

            # Store the model trained on the largest sample size for single image testing
            if s_size == sample_sizes[-1]: # If this is the run with the largest specified sample size
                 if model_name_str not in best_models_dict or s_size > best_models_dict[model_name_str]['sample_size_trained_on']:
                    best_models_dict[model_name_str] = {
                        'model_instance': final_model_to_eval, # Already fitted by evaluate_model
                        'sample_size_trained_on': s_size,
                        'is_quantum': is_q_model # Store type for test_single_image
                    }

    if results_list:
        print("\n\n=== FINAL RESULTS SUMMARY ===")
        df_results = pd.DataFrame(results_list)
        print(df_results[['model', 'sample_size', 'accuracy', 'balanced_accuracy', 'f1_macro', 'train_time', 'predict_time']])

        plt.figure(figsize=(12, 6))
        for model_name_unique in df_results['model'].unique():
            subset_df = df_results[df_results['model'] == model_name_unique]
            plt.plot(subset_df['sample_size'], subset_df['accuracy'], 'o-', label=model_name_unique)
        plt.title("Model Accuracy vs. Training Sample Size")
        plt.xlabel("Training Sample Size")
        plt.ylabel("Test Accuracy")
        plt.legend()
        plt.grid(True)
        plt.show()

    if best_models_dict:
        print("\n\n=== TESTING BEST MODELS WITH SAMPLE IMAGES ===")
        # Construct paths to a few test images (e.g., first image from first 3 classes)
        sample_image_paths_to_test = []
        for cls_name_iter in classes[:3]:
            cls_folder_path = os.path.join(image_dir, cls_name_iter)
            if os.path.exists(cls_folder_path):
                images_in_folder = glob(os.path.join(cls_folder_path, "*.jpg"))
                if images_in_folder:
                    sample_image_paths_to_test.append(images_in_folder[0])

        for img_path_to_test in sample_image_paths_to_test:
            if os.path.exists(img_path_to_test):
                print(f"\n--- Testing image: {os.path.basename(img_path_to_test)} ---")
                for model_id_name_test, model_info_test in best_models_dict.items():
                    print(f"Model: {model_id_name_test} (trained on {model_info_test['sample_size_trained_on']} samples)")
                    model_type_for_test = 'quantum' if model_info_test['is_quantum'] else 'classical'
                    test_single_image(
                        model_info_test['model_instance'], # The fitted model
                        img_path_to_test,
                        scaler_fitted, # The scaler fitted on training CNN features
                        pca_fitted if model_type_for_test == 'quantum' else None, # PCA fitted on training scaled CNN features
                        model_type_str=model_type_for_test
                    )
            else:
                print(f"Sample image path not found: {img_path_to_test}")

# Run the experiment
if __name__ == "__main__":
    # Ensure matplotlib uses a non-interactive backend if not in Jupyter
    # or if issues arise with plt.show() blocking.
    # For script execution, sometimes 'Agg' is used.
    # If running in Colab/Jupyter, %matplotlib inline (or default) is usually fine.
    try:
        run_experiment()
    except Exception as e:
        print(f"Error in experiment: {e}")
        traceback.print_exc()
    finally:
        print("\n=== EXPERIMENT COMPLETE ===")
```
